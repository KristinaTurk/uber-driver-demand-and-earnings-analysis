# External data collection- Dubrovnik weather data

This notebook retrieves historical hourly weather data for Dubrovnik, Croatia from year 2024 to 2025.

Source: Open-Mateo.com API

Output: CSV file stored to 'uber_driver_project/data/raw/weather_hourly_2024_2025.csv'


## Importing libraries

```python
import requests
import pandas as pd
```

#### Dubrovnik coordinates

```python
latitude = 42.6481
longitude = 18.0922
```

## Configuration

```python
url = 'https://archive-api.open-meteo.com/v1/archive'
```

```python
params = {
    'latitude':42.6481,
    'longitude':18.0922,
    'start_date':'2024-04-19',
    'end_date':'2025-08-29',
    'hourly':'temperature_2m,precipitation,weathercode,apparent_temperature'
}
```

## API request

```python
response = requests.get(url,params=params)
```

```python
data = response.json()
```

## Convert JSON to Dataframe

```python
hourly_data = data['hourly']
df_weather = pd.DataFrame(hourly_data)
df_weather.head()
```

```python
df_weather.tail()
```

## Save DataFrame to CSV file

```python
df_weather.to_csv('../data/raw_data/weather_hourly_2024_2025.csv', index = False)
```
