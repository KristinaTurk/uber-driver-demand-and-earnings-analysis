# External data collection- Dubrovnik flight data
This notebook retrieves historical flight arrival and departure data for Dubrovnik Airport, from year 2024 to 2025.

## Data availability limitation
Historical flight arrival and departure data for Dubrovnik Airport was investigated as a potentially important external variable for Uber demand analysis.
However, most available APIs providing historical flight data required paid subscriptions or enterprise access, which was outside of the scope of this project.
Therefore, flight traffic data was not included in the current version of the analysis.
Future versions of the project could integrate historical flight data if a suitable open dataset or affordable API becomes available.
